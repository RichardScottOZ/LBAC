# LBAC - Late Bronze Age Collapse

A Diplomacy, Trading and Conflict Resource Management Game

Experience the tumultuous period of the Late Bronze Age Collapse (circa 1200 BC). Lead one of the great civilizations of the ancient world through unprecedented crisis, navigating diplomacy, trade, and conflict to ensure your people's survival.

## How to Play

Run the cell below to start the game. When prompted, type your choice and press **Enter**.

**Controls:**
- `1` - Diplomacy (gifts, alliances, threats)
- `2` - Trade (routes, exchange, produce bronze)
- `3` - Military (recruit, raid, war)
- `4` - Internal Affairs (agriculture, tech, festivals)
- `5` - View Status
- `6` - End Turn
- `7` - Quit

> **Tip:** If the game gets stuck, use **Kernel → Restart & Run All** to restart from scratch.
